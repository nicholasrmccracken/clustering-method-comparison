```python
# Import libraries
import pandas as pd
import numpy as np
from scipy.spatial import distance
from scipy.spatial.distance import cdist
from sklearn import cluster 
from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering
from sklearn.cluster import SpectralClustering
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import seaborn as sns
import time
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
from scipy.optimize import linear_sum_assignment
import warnings
warnings.filterwarnings('ignore')

#np.set_printoptions(threshold=np.inf)
np.set_printoptions(linewidth=120)
```

```python
def plot_clusters(title, X, y, size):
    # Plot scatterplot of clustered data.
    plt.figure(figsize=(size, size))
    plt.title(title)
    plt.xlabel("X0") 
    plt.ylabel("X1")
    plt.scatter(X[:, 0], X[:, 1], s = 40, c = y[:]) 
    plt.show()
    plt.clf()
```

```python
def remap_cluster_labels(true_y, pred_y):
    # Rearrange the predicted cluster labels so they best match the true ones.
    # Best remaps cluster labels to true labels before evaluating quality of the clustering algorithm.

    conf_mat = confusion_matrix(true_y, pred_y)  #SciKitLearn convention is rows are true labels, cols are predicted labels
    
    # Find the optimal remapping of labels, so the predicted labels best match the true labels.
    # col_index is the predicted label; row_index is the true label to change it to.
    # Negate the costs so maximizing the assignment minimizes the mismatches.
    row_index, col_index = linear_sum_assignment(conf_mat.max()-conf_mat)

    # Remap predicted labels to best match true labels
    new_pred_y = np.zeros(len(pred_y))
    for r in range(0,len(row_index)):
        old = col_index[r]  # the predicted y's
        new = row_index[r]  # remap to the corresponding true y's
        new_pred_y[pred_y == old] = new

    return new_pred_y 
```

```python
def compute_single_sse(df):
    # Compute the SSE for a single cluster stored in a dataframe with columns X0 and X1.
    cluster_points = df[['X0', 'X1']].to_numpy()
    cluster_center = np.mean(cluster_points, axis=0)
    SSE = np.sum((cluster_points - cluster_center) ** 2)
    return SSE

def compute_all_sse(df):  # Columns are X0, X1, y
    # Compute WSS, BSS, and total SSE for a dataframe with columns X0, X1, and y.
    X = df[['X0', 'X1']].to_numpy()
    Y = df['y'].to_numpy()
    
    sum_WSS = 0  # WSS is the sum of the SSE for each cluster
    for cluster_id in np.unique(Y):
        cluster_df = df[df['y'] == cluster_id]
        sum_WSS += compute_single_sse(cluster_df)
        
    center = np.mean(X, axis=0)
    total_SSE = np.sum((X - center) ** 2)
    
    BSS = total_SSE - sum_WSS
    
    return sum_WSS, BSS, total_SSE
```

```python
def print_evaluation_results(title_string, show_true, show_pred, X, true_y, pred_y, runtime):
    # Print clustering evaluation results.
    print(title_string)
    print()
    
    # True cluster metrics
    if show_true:
        # Build dataframe with true labels
        true_df = pd.DataFrame(X, columns=['X0', 'X1'])
        true_df['y'] = true_y
        
        true_wss, true_bss, true_total_sse = compute_all_sse(true_df)

        print("True Cluster Metrics")
        print(f"WSS: {true_wss:.4f}")
        print(f"BSS: {true_bss:.4f}")
        print(f"Total SSE: {true_total_sse:.4f}")
        print()

    # Predicted cluster metrics
    if pred_y is not None:
        # For SSE/WSS/BSS, use the predicted labels as given
        pred_df = pd.DataFrame(X, columns=['X0', 'X1'])
        pred_df['y'] = pred_y
        
        pred_wss, pred_bss, pred_total_sse = compute_all_sse(pred_df)

        print("Predicted Cluster Metrics")
        print(f"WSS: {pred_wss:.4f}")
        print(f"BSS: {pred_bss:.4f}")
        print(f"Total SSE: {pred_total_sse:.4f}")
        print()
        
        # If DBSCAN produced noise points, exclude them from label-based evaluation
        if -1 in pred_y:
            # Exclude noise points (pred_y == -1) from evaluation
            mask = pred_y != -1
            eval_true_y = true_y[mask]
            eval_pred_y = pred_y[mask]

            if len(eval_pred_y) > 0:
                remapped_eval_pred_y = remap_cluster_labels(eval_true_y, eval_pred_y)

                print("Confusion Matrix (excluding noise points)")
                print(confusion_matrix(eval_true_y, remapped_eval_pred_y))
                print()

                print("Classification Report (excluding noise points)")
                print(classification_report(eval_true_y, remapped_eval_pred_y, zero_division=0))
                print()

                print(f"Accuracy (excluding noise points): {accuracy_score(eval_true_y, remapped_eval_pred_y):.4f}")
                print(f"Noise Points: {np.sum(pred_y == -1)}")
                print(f"Running Time: {runtime:.4f} seconds")
                print()
            else:
                print("All points were labeled as noise, so classification metrics could not be computed.")
                print(f"Noise Points: {np.sum(pred_y == -1)}")
                print(f"Running Time: {runtime:.4f} seconds")
                print()

            # For plotting, keep original DBSCAN labels including noise
            plot_y = pred_y
            
        else:
            # Standard case for algorithms like K-Means
            remapped_pred_y = remap_cluster_labels(true_y, pred_y)

            print("Confusion Matrix")
            print(confusion_matrix(true_y, remapped_pred_y))
            print()

            print("Classification Report")
            print(classification_report(true_y, remapped_pred_y, zero_division=0))
            print()

            print(f"Accuracy: {accuracy_score(true_y, remapped_pred_y):.4f}")
            print(f"Running Time: {runtime:.4f} seconds")
            print()

            plot_y = remapped_pred_y

        # Visualize predicted clusters    
        if show_pred:
            plot_clusters("Predicted Clusters", X, plot_y, 5)
```

DBSCAN was configured using the following parameters:

- **eps**: defines the neighborhood radius used to determine if points are close/dense enough to form a cluster.
  - **eps = 1.5**: small neighborhood size to test strict density requirements; expected to produce many small clusters and/or label many points as noise.
  - **eps = 2.5**: medium neighborhood size that allows nearby points to form clusters while still maintaining reasonable separation between clusters.
  - **eps = 3.5**: large neighborhood size that groups more points together, which reduces fragmentation  the most, but increases risk of merging unique clusters.
- **min_samples = 5**: minimum number of points required to form a "dense" region; value of 5 is commonly used, providing good balance for cluster detection and noise avoidance.

These eps values were chosen based on the visual scale of Dataset2. The spread of points in the scatterplot suggests that meaningful clusters could be found within a range of roughly 2-4 eps. These values target this range.

Evaluation uses WSS, BSS, total SSE, classification metrics after label remapping, runtime, and visualizations.

```python
for eps in [1.5, 2.5, 3.5]:
    start_time = time.time()

    dbscan = DBSCAN(
        eps=eps,
        min_samples=5
    )

    pred_y = dbscan.fit_predict(X)

    end_time = time.time()
    runtime = end_time - start_time

    print_evaluation_results(
        f"Dataset2 - DBSCAN with eps={eps}",
        False,
        True,
        X,
        true_y,
        pred_y,
        runtime
    )
```
